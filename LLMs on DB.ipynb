```python
import mlflow
from mlflow import pyfunc
class BasicChatBot(pyfunc.PythonModel):
    def __init__(self, model_name):
        self.model_name=model_name
    def chatCompletionAPI(self, user_query):
        openai_client=OpenAI(
             api_key=DATABRICKS_TOKEN,
             base_url="https://adb-7405610296555497.17.azuredatabricks.net/ai-gateway/mlflow/v1"
        )
        response=openai_client.chat.completions.create(
            model=self.model_name,
           messages=[
        {
            "role":"system",
            "content": [
                {
                    "type":"text",
                    "text":"You are batman, the protector of gotham"
                }
            ]
        },
        {
            "role": "user",
            "content":[
                {
                    "type":"text",
                    "text":user_query
                }
            ]
        }
    ]
        )
        return response.choices[0].message.content
    def predict(self, context, data):
        user_query=data['user_query'].iloc[0]
        gpt_response=self.chatCompletionAPI(user_query)
        return gpt_response
```

```python
#save model
test_model=BasicChatBot(model_name='system.ai.gemma-3-12b')

from mlflow.models import infer_signature
import pandas as pd 
import mlflow
input_example=pd.DataFrame(
    [
        {'user_query': "how is Gotham city?"}
    ]
)

output_example=pd.DataFrame(
    [
        {
        'predictions': 'Gotham city is facing cloudy weather'
        }
    ]
)

signature=infer_signature(input_example, output_example)
model_path="basicChatBot4"
mlflow.pyfunc.save_model(path=model_path,
                         python_model=test_model,
                         signature=signature,
                         input_example=input_example)
```
